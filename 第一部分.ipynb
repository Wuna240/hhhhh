```python
'''
1.利用高德地图 API 来获取周围的 POI数据，并将这些数据保存到 Excel 文件中。
代码主要分为几个部分：请求 API、处理数据、翻页获取更多数据以及保存结果。'''

import requests
import time
import pandas as pd

url='https://restapi.amap.com/v5/place/around?key=201f374c242dff3444664c1082fb0634&types=050103|050123&location=113.35,23.09&radius=5000&page_size=25&page_num=1'

'''
这行定义了一个基本的 URL，它是发送给高德地图 API 的请求地址。以下是 URL 的各个参数的解释：
key=yourkey：需要替换为你自己的高德地图 API Key。
types=050103|050123：指定要搜索的兴趣点类型。050103 是餐饮服务，050123 是购物服务（具体类型可以参考高德 API 文档）。
location=113.35,23.09：指定搜索的中心位置，经纬度。
radius=5000：搜索半径，单位是米，这里是 5000 米（5公里）。
page_size=25：每一页返回的 POI 数量，这里是 25。
page_num=1：当前请求的页码，默认为 1。
'''

page=0
poi_all=pd.DataFrame()


while page<100:
    url_next=url.replace('page_num=1','page_num='+str(page))
    poi=requests.get(url_next)
    poi_text=eval(poi.text)
    if poi_text['count']=='0':
        break
    else:
        poi_d=pd.DataFrame(poi_text['pois'])
        poi_all=pd.concat([poi_all,poi_d])
    page+=1
    time.sleep(3)
    print(page)

'''
while page<100:：循环最多获取 100 页数据。
url_next=url.replace('page_num=1','page_num='+str(page))：根据当前的 page 变量，构建新的 URL，请求不同页的数据。
poi=requests.get(url_next)：发送请求，获取返回的 POI 数据。
poi_text=eval(poi.text)：将返回的 JSON 格式字符串转换为 Python 字典，使用 eval() 可以直接解析返回的文本为字典对象。（注意：eval() 在处理外部数据时有潜在安全风险，通常可以使用 json.loads() 来替代。）
if poi_text['count']=='0':：如果返回的 POI 数量为 0，则退出循环，停止请求更多数据。
else:：如果有 POI 数据，则将这些数据转换为 DataFrame，并将其与已有的 poi_all 合并。
poi_d=pd.DataFrame(poi_text['pois'])：将当前页的 POI 数据转换为 DataFrame。
poi_all=pd.concat([poi_all,poi_d])：将当前页的数据与已有数据拼接，形成一个大的 DataFrame。
page+=1：更新页码，进入下一页。
time.sleep(3)：每次请求之间暂停 3 秒，避免请求频繁造成限制。
print(page)：打印当前请求的页码，用来跟踪进度。
'''

poi_all.to_excel(r'C:\Users\11707\Desktop\vspython\高德周边.xlsx','Sheet1',index=False)
  
```
