```python
# 3.利用folium将转换后的点显示在地图上。地图上画出不同距离线.

import pandas as pd
import folium
```

```python
# 读取 Excel 文件中的坐标,将每个坐标字符串按逗号分割成列表 [经度, 纬度]。最终 poi 是一个列表，每个元素是 [经度, 纬度]

poi=pd.read_excel(r'C:\Users\11707\Desktop\vspython\高德周边.xlsx','Sheet1')['newlocation'].str.split(',')  
```

```python
'''创建地图对象'''

poi_map = folium.Map(
    location=[23.090164,113.354053],
    zoom_start=10,
    tiles='http://webrd02.is.autonavi.com/appmaptile?lang=zh_cn&size=1&scale=1&style=8&x={x}&y={y}&z={z}&ltype=6',
    attr='AutoNavi'
)

'''
location=[23.090164,113.354053]: 地图中心位置，纬度在前，经度在后。
zoom_start=10: 地图缩放等级，10 是城市级别。
tiles=...: 使用高德地图的瓦片服务（地图底图）。
这是一个高德地图 Web 服务的 URL 模板，{x}, {y}, {z} 会被 folium 自动替换为瓦片坐标。
attr='AutoNavi': 地图来源说明（版权信息）。
'''
```

```python
'''在地图上添加单个标记点'''

folium.Marker(
    location=[23.0937830809255,113.35430088524312],
    icon=folium.Icon(icon="cloud", color="red")
).add_to(poi_map)
```

```python
'''循环添加所有 POI'''
for p in poi:
    folium.Marker(
        location=p[::-1],  # 经纬度颠倒
        icon=folium.Icon(icon="cloud", color="green")
    ).add_to(poi_map)
    
```

```python
'''添加同心圆'''
for c in range(1,6):
    folium.Circle(
        radius=1000*c,
        location=[23.0937830809255,113.35430088524312],
        color='crimson',
        fill=False,
    ).add_to(poi_map)

'''
range(1,6): 循环生成 5 个圆圈。
radius=1000*c: 半径分别是 1000、2000、3000、4000、5000 米。
location=[纬度, 经度]: 圆心位置。
color='crimson': 圆的边框颜色为红色。
fill=False: 圆形不填充，只有边框。
作用：在地图上标记某个中心点的不同距离范围，可用于分析 POI 的分布。
'''
```
